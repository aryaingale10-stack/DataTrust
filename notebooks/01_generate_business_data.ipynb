```python
# Email domains used in the synthetic customer dataset
EMAIL_DOMAINS = [
    "gmail.com",
    "yahoo.com",
    "outlook.com",
    "icloud.com"
]


def generate_email(first_name, last_name, customer_number):
    """
    Generate a unique email address for a customer.
    """
    domain = random.choice(EMAIL_DOMAINS)

    email = (
        f"{first_name.lower()}."
        f"{last_name.lower()}"
        f"{customer_number}@{domain}"
    )

    return email


def generate_phone():
    """
    Generate a synthetic 10-digit US-style phone number.
    """
    area_code = random.randint(200, 999)
    prefix = random.randint(200, 999)
    line_number = random.randint(1000, 9999)

    return f"{area_code}-{prefix}-{line_number}"
```

```python
# Create the working copy that will receive intentional data-quality issues
customers_dirty_df = customers_ground_truth_df.copy(deep=True)

# Store information about every intentionally injected issue
customer_issue_log = []

print("Dirty customer dataset initialized.")
print("Rows:", len(customers_dirty_df))
print("Issue log initialized:", len(customer_issue_log))
```

```python
# ---------------------------------------------------------
# Inject Issue 1: Missing Region Values
# ---------------------------------------------------------

missing_region_rate = 0.01

# Calculate number of rows to corrupt
num_missing_regions = int(
    len(customers_dirty_df) * missing_region_rate
)

# Randomly select rows
missing_region_indices = np.random.choice(
    customers_dirty_df.index,
    size=num_missing_regions,
    replace=False
)

# Log the issue BEFORE changing the data
for idx in missing_region_indices:

    customer_issue_log.append({
        "customer_id": customers_dirty_df.loc[idx, "customer_id"],
        "issue_type": "missing_value",
        "column": "region",
        "original_value": customers_dirty_df.loc[idx, "region"],
        "injected_value": None
    })

# Inject missing values
customers_dirty_df.loc[
    missing_region_indices,
    "region"
] = None

print("Missing region values injected:", num_missing_regions)
print(
    "Current missing regions:",
    customers_dirty_df["region"].isnull().sum()
)

print(
    "Total issues logged:",
    len(customer_issue_log)
)
```

```python
customers_dirty_df[
    customers_dirty_df["region"].isnull()
].head(10)
```

```python
# ---------------------------------------------------------
# Inject Issue 2: Invalid Email Addresses
# ---------------------------------------------------------

invalid_email_rate = 0.008

num_invalid_emails = int(
    len(customers_dirty_df) * invalid_email_rate
)

# Avoid customers already used for missing-region injection
available_indices = customers_dirty_df.index.difference(
    missing_region_indices
)

# Randomly select rows
invalid_email_indices = np.random.choice(
    available_indices,
    size=num_invalid_emails,
    replace=False
)

# Log original values BEFORE modifying them
for idx in invalid_email_indices:

    original_email = customers_dirty_df.loc[idx, "email"]

    # Remove @ to deliberately create an invalid email
    invalid_email = original_email.replace("@", "")

    customer_issue_log.append({
        "customer_id": customers_dirty_df.loc[idx, "customer_id"],
        "issue_type": "invalid_format",
        "column": "email",
        "original_value": original_email,
        "injected_value": invalid_email
    })

    # Inject corrupted email
    customers_dirty_df.loc[idx, "email"] = invalid_email


print("Invalid emails injected:", num_invalid_emails)

print(
    "Total issues logged:",
    len(customer_issue_log)
)
```

```python
customers_dirty_df.loc[
    invalid_email_indices,
    ["customer_id", "email"]
].head(10)
```

```python
# ---------------------------------------------------------
# Inject Issue 3: Geographic Region Mismatch
# ---------------------------------------------------------

region_mismatch_rate = 0.006

num_region_mismatches = int(
    len(customers_dirty_df) * region_mismatch_rate
)

# Exclude rows already corrupted
used_indices = set(missing_region_indices) | set(invalid_email_indices)

available_indices = customers_dirty_df.index[
    ~customers_dirty_df.index.isin(used_indices)
]

# Randomly select rows
region_mismatch_indices = np.random.choice(
    available_indices,
    size=num_region_mismatches,
    replace=False
)

# Inject incorrect regions
for idx in region_mismatch_indices:

    original_region = customers_dirty_df.loc[idx, "region"]

    # Select a region different from the correct one
    incorrect_regions = [
        region for region in REGIONS
        if region != original_region
    ]

    incorrect_region = random.choice(incorrect_regions)

    # Log the issue
    customer_issue_log.append({
        "customer_id": customers_dirty_df.loc[idx, "customer_id"],
        "issue_type": "consistency_error",
        "column": "region",
        "original_value": original_region,
        "injected_value": incorrect_region
    })

    # Replace correct region with incorrect region
    customers_dirty_df.loc[idx, "region"] = incorrect_region


print("Region mismatches injected:", num_region_mismatches)

print(
    "Total issues logged:",
    len(customer_issue_log)
)
```

```python
customers_dirty_df.loc[
    region_mismatch_indices,
    ["customer_id", "city", "state", "region"]
].head(10)
```

```python
# ---------------------------------------------------------
# Inject Issue 4: Duplicate Customer Records
# ---------------------------------------------------------

num_duplicate_customers = 200

# Select records that have not already been corrupted
used_indices = (
    set(missing_region_indices)
    | set(invalid_email_indices)
    | set(region_mismatch_indices)
)

available_indices = customers_dirty_df.index[
    ~customers_dirty_df.index.isin(used_indices)
]

# Randomly select customers to duplicate
duplicate_source_indices = np.random.choice(
    available_indices,
    size=num_duplicate_customers,
    replace=False
)

# Copy selected customer records
duplicate_records = customers_dirty_df.loc[
    duplicate_source_indices
].copy()

# Log each duplicate
for idx in duplicate_source_indices:

    customer_issue_log.append({
        "customer_id": customers_dirty_df.loc[idx, "customer_id"],
        "issue_type": "duplicate_record",
        "column": "customer_id",
        "original_value": customers_dirty_df.loc[idx, "customer_id"],
        "injected_value": customers_dirty_df.loc[idx, "customer_id"]
    })

# Append duplicates to the dirty dataset
customers_dirty_df = pd.concat(
    [customers_dirty_df, duplicate_records],
    ignore_index=True
)

print("Duplicate customer records injected:", num_duplicate_customers)
print("New customer dataset size:", len(customers_dirty_df))
print("Total issues logged:", len(customer_issue_log))
```

```python
# ---------------------------------------------------------
# Generate Clean Product Dataset
# ---------------------------------------------------------

products = []

for i in range(1, NUM_PRODUCTS + 1):

    # Unique product ID
    product_id = f"P{i:05d}"

    # Select category
    category = random.choice(list(PRODUCT_CATEGORIES.keys()))

    # Select valid subcategory belonging to that category
    subcategory = random.choice(
        PRODUCT_CATEGORIES[category]
    )

    # Select brand belonging to that category
    brand = random.choice(
        BRANDS_BY_CATEGORY[category]
    )

    # Generate product name
    product_name = f"{brand} {subcategory} {i:04d}"

    # Get pricing configuration
    pricing = PRODUCT_PRICE_RANGES[category]

    # Generate product cost
    unit_cost = round(
        random.uniform(
            pricing["cost_min"],
            pricing["cost_max"]
        ),
        2
    )

    # Generate markup
    markup = random.uniform(
        pricing["markup_min"],
        pricing["markup_max"]
    )

    # Selling price
    unit_price = round(
        unit_cost * markup,
        2
    )

    # Supplier
    supplier_id = f"S{random.randint(1, 50):03d}"

    # Product launch date
    start_date = pd.Timestamp("2019-01-01")
    end_date = pd.Timestamp("2025-12-31")

    launch_date = start_date + pd.to_timedelta(
        random.randint(
            0,
            (end_date - start_date).days
        ),
        unit="D"
    )

    # Product status
    product_status = random.choices(
        ["Active", "Discontinued"],
        weights=[0.90, 0.10],
        k=1
    )[0]

    products.append({
        "product_id": product_id,
        "product_name": product_name,
        "category": category,
        "subcategory": subcategory,
        "brand": brand,
        "unit_cost": unit_cost,
        "unit_price": unit_price,
        "supplier_id": supplier_id,
        "launch_date": launch_date,
        "product_status": product_status
    })


# Convert to DataFrame
products_clean_df = pd.DataFrame(products)

print("Product dataset generated successfully!")
print("Number of rows:", len(products_clean_df))
print("Number of columns:", len(products_clean_df.columns))

products_clean_df.head()
```

```python
# ---------------------------------------------------------
# Inject Product Issue 1: Missing Subcategory
# ---------------------------------------------------------

missing_subcategory_rate = 0.01

# Calculate number of products to corrupt
num_missing_subcategories = int(
    len(products_dirty_df) * missing_subcategory_rate
)

# Randomly select product rows
missing_subcategory_indices = np.random.choice(
    products_dirty_df.index,
    size=num_missing_subcategories,
    replace=False
)

# Log each issue BEFORE changing the data
for idx in missing_subcategory_indices:

    product_issue_log.append({
        "product_id": products_dirty_df.loc[idx, "product_id"],
        "issue_type": "missing_value",
        "column": "subcategory",
        "original_value": products_dirty_df.loc[idx, "subcategory"],
        "injected_value": None
    })

# Inject missing values
products_dirty_df.loc[
    missing_subcategory_indices,
    "subcategory"
] = None


print("Missing subcategories injected:", num_missing_subcategories)

print(
    "Current missing subcategories:",
    products_dirty_df["subcategory"].isnull().sum()
)

print(
    "Total product issues logged:",
    len(product_issue_log)
)
```

```python
# ---------------------------------------------------------
# Inject Product Issue 2: Selling Price Below Unit Cost
# ---------------------------------------------------------

num_price_below_cost = 15

# Avoid products already corrupted
used_indices = set(missing_subcategory_indices)

available_indices = products_dirty_df.index[
    ~products_dirty_df.index.isin(used_indices)
]

# Randomly select products
price_below_cost_indices = np.random.choice(
    available_indices,
    size=num_price_below_cost,
    replace=False
)

# Inject incorrect prices
for idx in price_below_cost_indices:

    original_price = products_dirty_df.loc[idx, "unit_price"]
    unit_cost = products_dirty_df.loc[idx, "unit_cost"]

    # Generate selling price between 60% and 90% of cost
    invalid_price = round(
        unit_cost * random.uniform(0.60, 0.90),
        2
    )

    # Log issue before modifying data
    product_issue_log.append({
        "product_id": products_dirty_df.loc[idx, "product_id"],
        "issue_type": "business_rule_violation",
        "column": "unit_price",
        "original_value": original_price,
        "injected_value": invalid_price
    })

    # Inject incorrect price
    products_dirty_df.loc[idx, "unit_price"] = invalid_price


print("Price-below-cost issues injected:", num_price_below_cost)

print(
    "Products currently priced below cost:",
    (
        products_dirty_df["unit_price"]
        < products_dirty_df["unit_cost"]
    ).sum()
)

print(
    "Total product issues logged:",
    len(product_issue_log)
)
```

```python
# ---------------------------------------------------------
# Inject Product Issue 3: Category/Subcategory Mismatch
# ---------------------------------------------------------

num_category_mismatches = 15

# Avoid products already corrupted
used_indices = (
    set(missing_subcategory_indices)
    | set(price_below_cost_indices)
)

available_indices = products_dirty_df.index[
    ~products_dirty_df.index.isin(used_indices)
]

# Randomly select products
category_mismatch_indices = np.random.choice(
    available_indices,
    size=num_category_mismatches,
    replace=False
)

for idx in category_mismatch_indices:

    original_category = products_dirty_df.loc[idx, "category"]
    original_subcategory = products_dirty_df.loc[idx, "subcategory"]

    # Select a different category
    other_categories = [
        category
        for category in PRODUCT_CATEGORIES.keys()
        if category != original_category
    ]

    wrong_category = random.choice(other_categories)

    # Select a subcategory belonging to that WRONG category
    invalid_subcategory = random.choice(
        PRODUCT_CATEGORIES[wrong_category]
    )

    # Log the issue
    product_issue_log.append({
        "product_id": products_dirty_df.loc[idx, "product_id"],
        "issue_type": "consistency_error",
        "column": "subcategory",
        "original_value": original_subcategory,
        "injected_value": invalid_subcategory
    })

    # Replace the correct subcategory
    products_dirty_df.loc[
        idx,
        "subcategory"
    ] = invalid_subcategory


print(
    "Category/subcategory mismatches injected:",
    num_category_mismatches
)

print(
    "Total product issues logged:",
    len(product_issue_log)
)
```

```python
# ---------------------------------------------------------
# Order Behavior Configuration
# ---------------------------------------------------------

# Transaction history period
ORDER_START_DATE = pd.Timestamp("2023-01-01")
ORDER_END_DATE = pd.Timestamp("2025-12-31")

# Sales channel probabilities
CHANNEL_WEIGHTS = [
    0.45,   # Website
    0.35,   # Mobile App
    0.20    # Physical Store
]

# Order status probabilities
STATUS_WEIGHTS = [
    0.92,   # Completed
    0.05,   # Cancelled
    0.03    # Returned
]

# Payment method probabilities
PAYMENT_WEIGHTS = [
    0.40,   # Credit Card
    0.25,   # Debit Card
    0.20,   # PayPal
    0.15    # Digital Wallet
]

print("Order behavior configuration created.")

print(
    "Order period:",
    ORDER_START_DATE.date(),
    "to",
    ORDER_END_DATE.date()
)

print("\nChannel weights:")
for channel, weight in zip(ORDER_CHANNELS, CHANNEL_WEIGHTS):
    print(f"{channel}: {weight:.0%}")

print("\nStatus weights:")
for status, weight in zip(ORDER_STATUSES, STATUS_WEIGHTS):
    print(f"{status}: {weight:.0%}")

print("\nPayment weights:")
for method, weight in zip(PAYMENT_METHODS, PAYMENT_WEIGHTS):
    print(f"{method}: {weight:.0%}")
```

```python
# ---------------------------------------------------------
# Generate Clean Order Dataset
# ---------------------------------------------------------

orders = []

# Customer lookup for faster access
customer_lookup = (
    customers_ground_truth_df
    .set_index("customer_id")
)

customer_ids = customers_ground_truth_df["customer_id"].tolist()

for i in range(1, NUM_ORDERS + 1):

    # Unique order ID
    order_id = f"O{i:07d}"

    # Select customer
    customer_id = random.choice(customer_ids)

    customer = customer_lookup.loc[customer_id]

    # Customer signup date
    signup_date = pd.Timestamp(customer["signup_date"])

    # Order cannot occur before:
    # 1. NovaRetail transaction-history start date
    # 2. Customer signup date
    earliest_order_date = max(
        ORDER_START_DATE,
        signup_date
    )

    # Generate valid order date
    available_days = (
        ORDER_END_DATE - earliest_order_date
    ).days

    order_date = (
        earliest_order_date
        + pd.to_timedelta(
            random.randint(0, available_days),
            unit="D"
        )
    )

    # Sales channel
    sales_channel = random.choices(
        ORDER_CHANNELS,
        weights=CHANNEL_WEIGHTS,
        k=1
    )[0]

    # Order status
    order_status = random.choices(
        ORDER_STATUSES,
        weights=STATUS_WEIGHTS,
        k=1
    )[0]

    # Payment method
    payment_method = random.choices(
        PAYMENT_METHODS,
        weights=PAYMENT_WEIGHTS,
        k=1
    )[0]

    # Use customer's location as shipping location
    shipping_city = customer["city"]
    shipping_state = customer["state"]
    shipping_region = customer["region"]

    orders.append({
        "order_id": order_id,
        "customer_id": customer_id,
        "order_date": order_date,
        "sales_channel": sales_channel,
        "order_status": order_status,
        "payment_method": payment_method,
        "shipping_city": shipping_city,
        "shipping_state": shipping_state,
        "shipping_region": shipping_region
    })


# Convert to DataFrame
orders_clean_df = pd.DataFrame(orders)

print("Order dataset generated successfully!")
print("Number of rows:", len(orders_clean_df))
print("Number of columns:", len(orders_clean_df.columns))

orders_clean_df.head()
```

```python
# ---------------------------------------------------------
# Calculate Clean Order Financial Totals
# ---------------------------------------------------------

# Work from the CLEAN ground-truth order items
order_financials = order_items_ground_truth_df.copy()

# Gross sales before discount
order_financials["gross_sales"] = (
    order_financials["quantity"]
    * order_financials["unit_price"]
)

# Discount amount
order_financials["discount_amount"] = (
    order_financials["gross_sales"]
    * order_financials["discount_pct"]
)

# Net sales after discount
order_financials["net_sales"] = (
    order_financials["gross_sales"]
    - order_financials["discount_amount"]
)

# Total product cost
order_financials["total_cost"] = (
    order_financials["quantity"]
    * order_financials["unit_cost"]
)

# Gross profit
order_financials["gross_profit"] = (
    order_financials["net_sales"]
    - order_financials["total_cost"]
)

print("Line-item financial metrics calculated!")

order_financials.head(10)
```

```python
# ---------------------------------------------------------
# Aggregate Financial Metrics by Order
# ---------------------------------------------------------

order_totals_df = (
    order_financials
    .groupby("order_id", as_index=False)
    .agg({
        "gross_sales": "sum",
        "discount_amount": "sum",
        "net_sales": "sum",
        "total_cost": "sum",
        "gross_profit": "sum"
    })
)

# Round financial values to cents
financial_columns = [
    "gross_sales",
    "discount_amount",
    "net_sales",
    "total_cost",
    "gross_profit"
]

order_totals_df[financial_columns] = (
    order_totals_df[financial_columns]
    .round(2)
)

print("Order-level financial totals created!")

print(
    "Orders with financial totals:",
    len(order_totals_df)
)

print(
    "Unique order IDs:",
    order_totals_df["order_id"].nunique()
)

order_totals_df.head(10)
```

```python
# ---------------------------------------------------------
# Prepare Clean Payment Source Data
# ---------------------------------------------------------

payment_source_df = (
    orders_ground_truth_df[
        [
            "order_id",
            "order_date",
            "order_status",
            "payment_method"
        ]
    ]
    .merge(
        order_totals_df[
            [
                "order_id",
                "net_sales"
            ]
        ],
        on="order_id",
        how="left"
    )
)

print("Payment source data prepared!")

print(
    "Rows:",
    len(payment_source_df)
)

print(
    "Missing order totals:",
    payment_source_df["net_sales"]
    .isnull()
    .sum()
)

payment_source_df.head()
```

```python
# ---------------------------------------------------------
# Generate Clean Payments Dataset
# ---------------------------------------------------------

payments = []

for i, row in payment_source_df.iterrows():

    payment_id = f"PAY{i + 1:07d}"

    # Map order status to payment status
    if row["order_status"] == "Completed":
        payment_status = "Successful"

    elif row["order_status"] == "Cancelled":
        payment_status = "Failed"

    else:
        payment_status = "Refunded"

    # Keep payment date aligned with the order date
    payment_date = row["order_date"]

    # Expected transaction amount
    payment_amount = round(
        float(row["net_sales"]),
        2
    )

    payments.append({
        "payment_id": payment_id,
        "order_id": row["order_id"],
        "payment_date": payment_date,
        "payment_method": row["payment_method"],
        "payment_status": payment_status,
        "payment_amount": payment_amount
    })


payments_clean_df = pd.DataFrame(payments)

print("Clean payments dataset created!")

print(
    "Total payment rows:",
    len(payments_clean_df)
)

print(
    "Total columns:",
    len(payments_clean_df.columns)
)

payments_clean_df.head(10)
```

```python
# ---------------------------------------------------------
# Initialize Dirty Payments Dataset
# ---------------------------------------------------------

payments_dirty_df = (
    payments_ground_truth_df.copy(deep=True)
)

payment_issue_log = []

print("Dirty payments dataset initialized!")

print(
    "Current rows:",
    len(payments_dirty_df)
)

print(
    "Current columns:",
    len(payments_dirty_df.columns)
)

print(
    "Current injected issues:",
    len(payment_issue_log)
)
```

```python
# ---------------------------------------------------------
# Inject Payment Issue 1: Payment Amount Mismatch
# ---------------------------------------------------------

num_payment_amount_mismatches = 200

# Randomly select payment rows
payment_amount_mismatch_indices = np.random.choice(
    payments_dirty_df.index,
    size=num_payment_amount_mismatches,
    replace=False
)

for idx in payment_amount_mismatch_indices:

    original_amount = float(
        payments_dirty_df.loc[idx, "payment_amount"]
    )

    # Change the payment by 10%-30%
    multiplier = random.choice([
        0.70,
        0.80,
        0.90,
        1.10,
        1.20,
        1.30
    ])

    incorrect_amount = round(
        original_amount * multiplier,
        2
    )

    # Log issue before modification
    payment_issue_log.append({
        "payment_id": payments_dirty_df.loc[
            idx,
            "payment_id"
        ],
        "issue_type": "reconciliation_error",
        "column": "payment_amount",
        "original_value": original_amount,
        "injected_value": incorrect_amount
    })

    # Inject incorrect payment amount
    payments_dirty_df.loc[
        idx,
        "payment_amount"
    ] = incorrect_amount


print(
    "Payment amount mismatches injected:",
    num_payment_amount_mismatches
)

print(
    "Total payment issues logged:",
    len(payment_issue_log)
)
```

```python
# ---------------------------------------------------------
# Verify Payment Amount Mismatches
# ---------------------------------------------------------

payment_amount_check = (
    payments_dirty_df.loc[
        payment_amount_mismatch_indices,
        [
            "payment_id",
            "order_id",
            "payment_amount"
        ]
    ]
    .merge(
        order_totals_df[
            [
                "order_id",
                "net_sales"
            ]
        ],
        on="order_id",
        how="left"
    )
)

payment_amount_check["difference"] = (
    payment_amount_check["payment_amount"]
    - payment_amount_check["net_sales"]
).round(2)

payment_amount_check["is_mismatch"] = (
    payment_amount_check["difference"].abs() > 0.01
)

print(
    "Confirmed payment amount mismatches:",
    payment_amount_check["is_mismatch"].sum()
)

payment_amount_check.head(10)
```

```python
# ---------------------------------------------------------
# Inject Payment Issue 2: Nonexistent Order IDs
# ---------------------------------------------------------

num_orphan_payment_orders = 150

# Avoid rows already corrupted
used_indices = set(
    payment_amount_mismatch_indices
)

available_indices = payments_dirty_df.index[
    ~payments_dirty_df.index.isin(used_indices)
]

# Randomly select rows
orphan_payment_order_indices = np.random.choice(
    available_indices,
    size=num_orphan_payment_orders,
    replace=False
)

# Inject nonexistent order IDs
for counter, idx in enumerate(
    orphan_payment_order_indices,
    start=1
):

    original_order_id = payments_dirty_df.loc[
        idx,
        "order_id"
    ]

    # Real order IDs only run through O0120000
    invalid_order_id = f"O{9000000 + counter:07d}"

    # Log issue before modification
    payment_issue_log.append({
        "payment_id": payments_dirty_df.loc[
            idx,
            "payment_id"
        ],
        "issue_type": "referential_integrity_error",
        "column": "order_id",
        "original_value": original_order_id,
        "injected_value": invalid_order_id
    })

    # Inject nonexistent order ID
    payments_dirty_df.loc[
        idx,
        "order_id"
    ] = invalid_order_id


print(
    "Nonexistent order IDs injected:",
    num_orphan_payment_orders
)

print(
    "Total payment issues logged:",
    len(payment_issue_log)
)
```

```python
# ---------------------------------------------------------
# Verify Nonexistent Order References
# ---------------------------------------------------------

valid_order_ids = set(
    orders_ground_truth_df["order_id"]
)

orphan_payment_mask = (
    payments_dirty_df["order_id"].notna()
    &
    ~payments_dirty_df["order_id"].isin(
        valid_order_ids
    )
)

print(
    "Payments referencing nonexistent orders:",
    orphan_payment_mask.sum()
)

payments_dirty_df.loc[
    orphan_payment_mask,
    [
        "payment_id",
        "order_id",
        "payment_date",
        "payment_method",
        "payment_amount"
    ]
].head(10)
```

```python
# ---------------------------------------------------------
# Inject Payment Issue 3: Payment Before Order Date
# ---------------------------------------------------------

num_payment_date_errors = 100

# Avoid rows already corrupted
used_indices = (
    set(payment_amount_mismatch_indices)
    | set(orphan_payment_order_indices)
)

available_indices = payments_dirty_df.index[
    ~payments_dirty_df.index.isin(used_indices)
]

# Randomly select payment rows
payment_date_error_indices = np.random.choice(
    available_indices,
    size=num_payment_date_errors,
    replace=False
)

for idx in payment_date_error_indices:

    original_payment_date = pd.Timestamp(
        payments_dirty_df.loc[idx, "payment_date"]
    )

    # Move payment 1-7 days before the order date
    days_before = random.randint(1, 7)

    invalid_payment_date = (
        original_payment_date
        - pd.to_timedelta(days_before, unit="D")
    )

    # Log issue before modification
    payment_issue_log.append({
        "payment_id": payments_dirty_df.loc[
            idx,
            "payment_id"
        ],
        "issue_type": "temporal_consistency_error",
        "column": "payment_date",
        "original_value": original_payment_date,
        "injected_value": invalid_payment_date
    })

    # Inject incorrect payment date
    payments_dirty_df.loc[
        idx,
        "payment_date"
    ] = invalid_payment_date


print(
    "Payment-date errors injected:",
    num_payment_date_errors
)

print(
    "Total payment issues logged:",
    len(payment_issue_log)
)
```

```python
# ---------------------------------------------------------
# Verify Payment-Date Inconsistencies
# ---------------------------------------------------------

payment_date_check = (
    payments_dirty_df.loc[
        payment_date_error_indices,
        [
            "payment_id",
            "order_id",
            "payment_date"
        ]
    ]
    .merge(
        orders_ground_truth_df[
            [
                "order_id",
                "order_date"
            ]
        ],
        on="order_id",
        how="left"
    )
)

payment_date_check["payment_before_order"] = (
    pd.to_datetime(payment_date_check["payment_date"])
    <
    pd.to_datetime(payment_date_check["order_date"])
)

print(
    "Confirmed payments before order date:",
    payment_date_check[
        "payment_before_order"
    ].sum()
)

payment_date_check.head(10)
```

```python
# ---------------------------------------------------------
# Inject Payment Issue 4: Payment Status Inconsistency
# ---------------------------------------------------------

num_payment_status_errors = 100

# Avoid rows already corrupted
used_indices = (
    set(payment_amount_mismatch_indices)
    | set(orphan_payment_order_indices)
    | set(payment_date_error_indices)
)

available_indices = payments_dirty_df.index[
    ~payments_dirty_df.index.isin(used_indices)
]

# Randomly select payment rows
payment_status_error_indices = np.random.choice(
    available_indices,
    size=num_payment_status_errors,
    replace=False
)

# Correct mapping used by our NovaRetail business rules
EXPECTED_PAYMENT_STATUS = {
    "Completed": "Successful",
    "Cancelled": "Failed",
    "Returned": "Refunded"
}

for idx in payment_status_error_indices:

    order_id = payments_dirty_df.loc[
        idx,
        "order_id"
    ]

    # Find corresponding order status
    order_status = orders_ground_truth_df.loc[
        orders_ground_truth_df["order_id"] == order_id,
        "order_status"
    ].iloc[0]

    correct_status = EXPECTED_PAYMENT_STATUS[
        order_status
    ]

    # Choose a status different from the correct one
    incorrect_statuses = [
        status
        for status in PAYMENT_STATUSES
        if status != correct_status
    ]

    incorrect_status = random.choice(
        incorrect_statuses
    )

    original_status = payments_dirty_df.loc[
        idx,
        "payment_status"
    ]

    # Log issue before modification
    payment_issue_log.append({
        "payment_id": payments_dirty_df.loc[
            idx,
            "payment_id"
        ],
        "issue_type": "consistency_error",
        "column": "payment_status",
        "original_value": original_status,
        "injected_value": incorrect_status
    })

    # Inject incorrect payment status
    payments_dirty_df.loc[
        idx,
        "payment_status"
    ] = incorrect_status


print(
    "Payment-status inconsistencies injected:",
    num_payment_status_errors
)

print(
    "Total payment issues logged:",
    len(payment_issue_log)
)
```

```python
# ---------------------------------------------------------
# Verify Payment-Status Inconsistencies
# ---------------------------------------------------------

payment_status_check = (
    payments_dirty_df.loc[
        payment_status_error_indices,
        [
            "payment_id",
            "order_id",
            "payment_status"
        ]
    ]
    .merge(
        orders_ground_truth_df[
            [
                "order_id",
                "order_status"
            ]
        ],
        on="order_id",
        how="left"
    )
)

payment_status_check["expected_payment_status"] = (
    payment_status_check["order_status"]
    .map(EXPECTED_PAYMENT_STATUS)
)

payment_status_check["status_mismatch"] = (
    payment_status_check["payment_status"]
    !=
    payment_status_check["expected_payment_status"]
)

print(
    "Confirmed payment-status inconsistencies:",
    payment_status_check["status_mismatch"].sum()
)

payment_status_check.head(10)
```

```python
# ---------------------------------------------------------
# Inject Payment Issue 5: Duplicate Payment Records
# ---------------------------------------------------------

num_duplicate_payments = 100

# Avoid rows already corrupted
used_indices = (
    set(payment_amount_mismatch_indices)
    | set(orphan_payment_order_indices)
    | set(payment_date_error_indices)
    | set(payment_status_error_indices)
)

available_indices = payments_dirty_df.index[
    ~payments_dirty_df.index.isin(used_indices)
]

# Randomly select clean payment records
duplicate_payment_source_indices = np.random.choice(
    available_indices,
    size=num_duplicate_payments,
    replace=False
)

# Copy selected payment records
duplicate_payment_records = payments_dirty_df.loc[
    duplicate_payment_source_indices
].copy()

# Log each duplicate issue
for idx in duplicate_payment_source_indices:

    payment_id = payments_dirty_df.loc[
        idx,
        "payment_id"
    ]

    payment_issue_log.append({
        "payment_id": payment_id,
        "issue_type": "duplicate_record",
        "column": "payment_id",
        "original_value": payment_id,
        "injected_value": payment_id
    })

# Append duplicate rows
payments_dirty_df = pd.concat(
    [
        payments_dirty_df,
        duplicate_payment_records
    ],
    ignore_index=True
)

print(
    "Duplicate payments injected:",
    num_duplicate_payments
)

print(
    "New payments dataset size:",
    len(payments_dirty_df)
)

print(
    "Total payment issues logged:",
    len(payment_issue_log)
)
```

```python
# ---------------------------------------------------------
# Inject Return Issue 1: Invalid Return Quantity
# ---------------------------------------------------------

num_invalid_return_quantities = 100

# Randomly select return records
invalid_return_quantity_indices = np.random.choice(
    returns_dirty_df.index,
    size=num_invalid_return_quantities,
    replace=False
)

# Lookup purchased quantities by order_item_id
purchased_quantity_lookup = (
    order_items_ground_truth_df
    .set_index("order_item_id")["quantity"]
)

for idx in invalid_return_quantity_indices:

    order_item_id = returns_dirty_df.loc[
        idx,
        "order_item_id"
    ]

    original_return_quantity = returns_dirty_df.loc[
        idx,
        "return_quantity"
    ]

    purchased_quantity = int(
        purchased_quantity_lookup.loc[
            order_item_id
        ]
    )

    # Force returned quantity to exceed purchased quantity
    invalid_quantity = purchased_quantity + np.random.randint(
        1,
        4
    )

    returns_dirty_df.loc[
        idx,
        "return_quantity"
    ] = invalid_quantity

    # Record the intentionally injected issue
    return_issue_log.append({
        "return_id": returns_dirty_df.loc[idx, "return_id"],
        "issue_type": "business_rule_violation",
        "column": "return_quantity",
        "original_value": original_return_quantity,
        "injected_value": invalid_quantity
    })


print(
    "Invalid return quantities injected:",
    num_invalid_return_quantities
)

print(
    "Total return issues logged:",
    len(return_issue_log)
)
```

```python
# ---------------------------------------------------------
# Financial Reconciliation Audit 1:
# Order Totals -> Payments
# ---------------------------------------------------------

# Recalculate expected order totals from CLEAN order-item ground truth
expected_order_totals = (
    order_items_ground_truth_df
    .assign(
        expected_net_sales=lambda df:
            df["quantity"]
            * df["unit_price"]
            * (1 - df["discount_pct"])
    )
    .groupby(
        "order_id",
        as_index=False
    )["expected_net_sales"]
    .sum()
)

expected_order_totals[
    "expected_net_sales"
] = expected_order_totals[
    "expected_net_sales"
].round(2)

print(
    "Expected order totals calculated:",
    len(expected_order_totals)
)
```

```python
# ---------------------------------------------------------
# Prepare Payments for Reconciliation
# ---------------------------------------------------------

# Remove orphan payments
valid_payments_for_reconciliation = audit_payments[
    audit_payments["order_id"].isin(
        expected_order_totals["order_id"]
    )
].copy()

# Remove ALL occurrences of duplicated payment IDs
# so duplicate problems remain a separate quality test
duplicate_payment_ids = set(
    audit_payments.loc[
        audit_payments["payment_id"].duplicated(
            keep=False
        ),
        "payment_id"
    ]
)

valid_payments_for_reconciliation = (
    valid_payments_for_reconciliation[
        ~valid_payments_for_reconciliation[
            "payment_id"
        ].isin(duplicate_payment_ids)
    ]
    .copy()
)

print(
    "Payments eligible for amount reconciliation:",
    len(valid_payments_for_reconciliation)
)

print(
    "Duplicate payment IDs excluded:",
    len(duplicate_payment_ids)
)

print(
    "Orphan payments excluded:",
    len(payments_orphan_order)
)
```

```python
# ---------------------------------------------------------
# Detect Payment Amount Reconciliation Errors
# ---------------------------------------------------------

payment_reconciliation = (
    valid_payments_for_reconciliation.merge(
        expected_order_totals,
        on="order_id",
        how="inner"
    )
)

payment_reconciliation[
    "amount_difference"
] = (
    payment_reconciliation["payment_amount"]
    - payment_reconciliation["expected_net_sales"]
).round(2)

payment_amount_mismatches = payment_reconciliation[
    payment_reconciliation[
        "amount_difference"
    ].abs() > 0.01
]

print("PAYMENT RECONCILIATION RESULTS")
print("------------------------------")

print(
    "Payments checked:",
    len(payment_reconciliation)
)

print(
    "Payment amount mismatches:",
    len(payment_amount_mismatches)
)

display(
    payment_amount_mismatches[
        [
            "payment_id",
            "order_id",
            "payment_amount",
            "expected_net_sales",
            "amount_difference"
        ]
    ].head(10)
)
```

```python
# ---------------------------------------------------------
# Financial Reconciliation Audit 2:
# Returns -> Expected Refund Amount
# ---------------------------------------------------------

# Original transaction information
return_transaction_lookup = (
    order_items_ground_truth_df[
        [
            "order_item_id",
            "unit_price",
            "discount_pct",
            "quantity"
        ]
    ]
    .copy()
)

# Identify duplicated return IDs
duplicate_return_ids = set(
    audit_returns.loc[
        audit_returns["return_id"].duplicated(
            keep=False
        ),
        "return_id"
    ]
)

print(
    "Duplicate return IDs identified:",
    len(duplicate_return_ids)
)
```

```python
# ---------------------------------------------------------
# Prepare Returns for Refund Reconciliation
# ---------------------------------------------------------

returns_for_reconciliation = audit_returns[
    ~audit_returns["return_id"].isin(
        duplicate_return_ids
    )
].copy()

returns_for_reconciliation = (
    returns_for_reconciliation.merge(
        return_transaction_lookup,
        on="order_item_id",
        how="inner"
    )
)

# Exclude records where return quantity exceeds
# the originally purchased quantity
returns_for_reconciliation = (
    returns_for_reconciliation[
        (
            returns_for_reconciliation["return_quantity"] > 0
        )
        &
        (
            returns_for_reconciliation["return_quantity"]
            <= returns_for_reconciliation["quantity"]
        )
    ]
    .copy()
)

print(
    "Returns eligible for refund reconciliation:",
    len(returns_for_reconciliation)
)

print(
    "Duplicate return IDs excluded:",
    len(duplicate_return_ids)
)

print(
    "Orphan order-item references excluded:",
    len(returns_orphan_order_item)
)
```

```python
# ---------------------------------------------------------
# Calculate Expected Refund Amount
# ---------------------------------------------------------

returns_for_reconciliation[
    "expected_refund_amount"
] = (
    returns_for_reconciliation["return_quantity"]
    * returns_for_reconciliation["unit_price"]
    * (
        1
        - returns_for_reconciliation["discount_pct"]
    )
).round(2)

returns_for_reconciliation[
    "refund_difference"
] = (
    returns_for_reconciliation["refund_amount"]
    - returns_for_reconciliation["expected_refund_amount"]
).round(2)

refund_amount_mismatches = (
    returns_for_reconciliation[
        returns_for_reconciliation[
            "refund_difference"
        ].abs() > 0.01
    ]
)

print("RETURN REFUND RECONCILIATION RESULTS")
print("------------------------------------")

print(
    "Returns checked:",
    len(returns_for_reconciliation)
)

print(
    "Refund amount mismatches:",
    len(refund_amount_mismatches)
)

display(
    refund_amount_mismatches[
        [
            "return_id",
            "order_item_id",
            "return_quantity",
            "unit_price",
            "discount_pct",
            "refund_amount",
            "expected_refund_amount",
            "refund_difference"
        ]
    ].head(10)
)
```

```python
# ---------------------------------------------------------
# Temporal Audit 1:
# Orders Before Customer Signup
# ---------------------------------------------------------

# Use clean customer master to obtain the expected signup date
customer_signup_lookup = (
    customers_ground_truth_df[
        [
            "customer_id",
            "signup_date"
        ]
    ]
    .copy()
)

# Start from orders with valid customer IDs
orders_for_temporal_check = audit_orders[
    audit_orders["customer_id"].isin(
        customer_signup_lookup["customer_id"]
    )
].copy()

# Remove duplicate order IDs so duplicate records
# remain a separate quality issue
orders_for_temporal_check = (
    orders_for_temporal_check
    .drop_duplicates(
        subset=["order_id"],
        keep="first"
    )
)

# Join customer signup dates
orders_for_temporal_check = (
    orders_for_temporal_check.merge(
        customer_signup_lookup,
        on="customer_id",
        how="left"
    )
)

# Convert to datetime
orders_for_temporal_check["order_date"] = pd.to_datetime(
    orders_for_temporal_check["order_date"]
)

orders_for_temporal_check["signup_date"] = pd.to_datetime(
    orders_for_temporal_check["signup_date"]
)

print(
    "Orders eligible for temporal check:",
    len(orders_for_temporal_check)
)
```

```python
# ---------------------------------------------------------
# Detect Orders Before Customer Signup
# ---------------------------------------------------------

orders_before_signup = orders_for_temporal_check[
    orders_for_temporal_check["order_date"]
    <
    orders_for_temporal_check["signup_date"]
].copy()

orders_before_signup[
    "days_before_signup"
] = (
    orders_before_signup["signup_date"]
    - orders_before_signup["order_date"]
).dt.days

print("ORDER TEMPORAL CONSISTENCY RESULTS")
print("----------------------------------")

print(
    "Orders before customer signup:",
    len(orders_before_signup)
)

display(
    orders_before_signup[
        [
            "order_id",
            "customer_id",
            "signup_date",
            "order_date",
            "days_before_signup"
        ]
    ].head(10)
)
```

```python
# ---------------------------------------------------------
# Temporal Audit 2:
# Payments Before Orders
# ---------------------------------------------------------

# Create clean order-date lookup
order_date_lookup_df = (
    orders_ground_truth_df[
        [
            "order_id",
            "order_date"
        ]
    ]
    .copy()
)

# Exclude payments whose order_id does not exist
payments_for_temporal_check = audit_payments[
    audit_payments["order_id"].isin(
        order_date_lookup_df["order_id"]
    )
].copy()

# Remove duplicate payment IDs so duplicates
# remain a separate quality issue
payments_for_temporal_check = (
    payments_for_temporal_check[
        ~payments_for_temporal_check[
            "payment_id"
        ].isin(duplicate_payment_ids)
    ]
    .copy()
)

# Bring in original order date
payments_for_temporal_check = (
    payments_for_temporal_check.merge(
        order_date_lookup_df,
        on="order_id",
        how="left"
    )
)

# Convert dates
payments_for_temporal_check["payment_date"] = pd.to_datetime(
    payments_for_temporal_check["payment_date"]
)

payments_for_temporal_check["order_date"] = pd.to_datetime(
    payments_for_temporal_check["order_date"]
)

print(
    "Payments eligible for temporal check:",
    len(payments_for_temporal_check)
)
```

```python
# ---------------------------------------------------------
# Detect Payments Before Order Date
# ---------------------------------------------------------

payments_before_order = payments_for_temporal_check[
    payments_for_temporal_check["payment_date"]
    <
    payments_for_temporal_check["order_date"]
].copy()

payments_before_order[
    "days_before_order"
] = (
    payments_before_order["order_date"]
    - payments_before_order["payment_date"]
).dt.days

print("PAYMENT TEMPORAL CONSISTENCY RESULTS")
print("------------------------------------")

print(
    "Payments before order date:",
    len(payments_before_order)
)

display(
    payments_before_order[
        [
            "payment_id",
            "order_id",
            "order_date",
            "payment_date",
            "days_before_order"
        ]
    ].head(10)
)
```

```python
# ---------------------------------------------------------
# Temporal Audit 3:
# Returns Outside 30-Day Window
# ---------------------------------------------------------

# Clean order-date lookup
return_order_date_lookup = (
    orders_ground_truth_df[
        [
            "order_id",
            "order_date"
        ]
    ]
    .copy()
)

# Start with returns whose order_id exists
returns_for_temporal_check = audit_returns[
    audit_returns["order_id"].isin(
        return_order_date_lookup["order_id"]
    )
].copy()

# Remove duplicate return IDs
# Duplicate records are handled by a separate quality rule
returns_for_temporal_check = (
    returns_for_temporal_check[
        ~returns_for_temporal_check[
            "return_id"
        ].isin(duplicate_return_ids)
    ]
    .copy()
)

# Bring in the original order date
returns_for_temporal_check = (
    returns_for_temporal_check.merge(
        return_order_date_lookup,
        on="order_id",
        how="left"
    )
)

# Convert dates
returns_for_temporal_check["return_date"] = pd.to_datetime(
    returns_for_temporal_check["return_date"]
)

returns_for_temporal_check["order_date"] = pd.to_datetime(
    returns_for_temporal_check["order_date"]
)

# Calculate number of days between order and return
returns_for_temporal_check[
    "days_to_return"
] = (
    returns_for_temporal_check["return_date"]
    - returns_for_temporal_check["order_date"]
).dt.days

print(
    "Returns eligible for temporal check:",
    len(returns_for_temporal_check)
)
```

```python
# ---------------------------------------------------------
# Detect Returns Outside Allowed Window
# ---------------------------------------------------------

returns_outside_window = returns_for_temporal_check[
    (
        returns_for_temporal_check["days_to_return"]
        < MIN_RETURN_DAYS
    )
    |
    (
        returns_for_temporal_check["days_to_return"]
        > MAX_RETURN_DAYS
    )
].copy()

print("RETURN TEMPORAL CONSISTENCY RESULTS")
print("-----------------------------------")

print(
    "Returns outside allowed window:",
    len(returns_outside_window)
)

print(
    "Minimum days among invalid returns:",
    returns_outside_window[
        "days_to_return"
    ].min()
)

print(
    "Maximum days among invalid returns:",
    returns_outside_window[
        "days_to_return"
    ].max()
)

display(
    returns_outside_window[
        [
            "return_id",
            "order_id",
            "order_date",
            "return_date",
            "days_to_return"
        ]
    ].head(10)
)
```

```python
# ---------------------------------------------------------
# Final Stage 2 Audit:
# Primary Identifier Uniqueness
# ---------------------------------------------------------

uniqueness_checks = {
    "customers": (
        audit_customers,
        "customer_id"
    ),
    "products": (
        audit_products,
        "product_id"
    ),
    "orders": (
        audit_orders,
        "order_id"
    ),
    "order_items": (
        audit_order_items,
        "order_item_id"
    ),
    "payments": (
        audit_payments,
        "payment_id"
    ),
    "returns": (
        audit_returns,
        "return_id"
    )
}

uniqueness_results = []

for dataset_name, (df, id_column) in uniqueness_checks.items():

    duplicate_rows = df[
        id_column
    ].duplicated().sum()

    duplicate_ids = df.loc[
        df[id_column].duplicated(
            keep=False
        ),
        id_column
    ].nunique()

    uniqueness_results.append({
        "dataset": dataset_name,
        "id_column": id_column,
        "total_rows": len(df),
        "unique_ids": df[id_column].nunique(),
        "duplicate_rows": duplicate_rows,
        "duplicate_ids": duplicate_ids
    })


uniqueness_audit_df = pd.DataFrame(
    uniqueness_results
)

display(uniqueness_audit_df)
```

```python
# ---------------------------------------------------------
# Step 133A:
# Consolidated Numeric Range Audit
# ---------------------------------------------------------

numeric_rules = {
    "products": {
        "unit_cost": {"min": 0},
        "unit_price": {"min": 0}
    },

    "order_items": {
        "quantity": {"min": 1},
        "unit_price": {"min": 0},
        "discount_pct": {"min": 0, "max": 0.30}
    },

    "payments": {
        "payment_amount": {"min": 0}
    },

    "returns": {
        "return_quantity": {"min": 1},
        "refund_amount": {"min": 0}
    }
}

numeric_range_results = []

for dataset_name, column_rules in numeric_rules.items():

    df_data = datasets[dataset_name]

    for column, rules in column_rules.items():

        invalid_mask = pd.Series(
            False,
            index=df_data.index
        )

        if "min" in rules:
            invalid_mask |= (
                df_data[column] < rules["min"]
            )

        if "max" in rules:
            invalid_mask |= (
                df_data[column] > rules["max"]
            )

        invalid_count = invalid_mask.sum()

        numeric_range_results.append({
            "dataset": dataset_name,
            "column": column,
            "invalid_count": int(invalid_count),
            "minimum_observed": df_data[column].min(),
            "maximum_observed": df_data[column].max(),
            "allowed_min": rules.get("min"),
            "allowed_max": rules.get("max")
        })

numeric_range_audit_df = pd.DataFrame(
    numeric_range_results
)

print("NUMERIC RANGE AUDIT")
print("-------------------")

display(numeric_range_audit_df)
```

```python
# ---------------------------------------------------------
# Step 136A:
# Prepare Order vs Customer Signup Temporal Check
# ---------------------------------------------------------

orders_temporal = audit_orders.copy()
customers_temporal = audit_customers.copy()

# Convert dates safely
orders_temporal["order_date"] = pd.to_datetime(
    orders_temporal["order_date"],
    errors="coerce"
)

customers_temporal["signup_date"] = pd.to_datetime(
    customers_temporal["signup_date"],
    errors="coerce"
)

# Use one customer record per ID so intentional
# customer duplicates do not multiply order rows
customer_signup_lookup = (
    customers_temporal[
        ["customer_id", "signup_date"]
    ]
    .dropna(subset=["customer_id"])
    .drop_duplicates(
        subset=["customer_id"],
        keep="first"
    )
)

orders_signup_check = (
    orders_temporal.merge(
        customer_signup_lookup,
        on="customer_id",
        how="inner"
    )
)

print(
    "Orders available for temporal check:",
    len(orders_signup_check)
)
```

```python
# ---------------------------------------------------------
# Step 137A:
# Prepare Payment vs Order Temporal Check
# ---------------------------------------------------------

payments_temporal = audit_payments.copy()
orders_payment_temporal = audit_orders.copy()

payments_temporal["payment_date"] = pd.to_datetime(
    payments_temporal["payment_date"],
    errors="coerce"
)

orders_payment_temporal["order_date"] = pd.to_datetime(
    orders_payment_temporal["order_date"],
    errors="coerce"
)

# Use one order record per order_id so intentional
# duplicate orders do not multiply payment records
order_date_lookup = (
    orders_payment_temporal[
        ["order_id", "order_date"]
    ]
    .dropna(subset=["order_id"])
    .drop_duplicates(
        subset=["order_id"],
        keep="first"
    )
)

payments_date_check = (
    payments_temporal.merge(
        order_date_lookup,
        on="order_id",
        how="inner"
    )
)

print(
    "Payments available for temporal check:",
    len(payments_date_check)
)
```

```python
# ---------------------------------------------------------
# Step 145:
# Diagnose Return-Reason Detection Discrepancy
# ---------------------------------------------------------

# Derive the valid return reasons directly from clean data
valid_return_reasons = set(
    returns_ground_truth_df[
        "return_reason"
    ].dropna().unique()
)

# Get all return_reason issues recorded in the issue log
return_reason_ground_truth = (
    return_issue_log_df[
        return_issue_log_df["column"]
        .astype(str)
        .str.lower()
        .eq("return_reason")
    ]
    .copy()
)

# Compare those exact return IDs with the current dirty dataset
return_reason_diagnostic = (
    return_reason_ground_truth.merge(
        audit_returns[
            ["return_id", "return_reason"]
        ],
        on="return_id",
        how="left"
    )
)

# Determine whether each injected issue is currently detectable
return_reason_diagnostic["currently_invalid"] = (
    return_reason_diagnostic["return_reason"].notna()
    &
    ~return_reason_diagnostic[
        "return_reason"
    ].isin(valid_return_reasons)
)

print("RETURN REASON DIAGNOSTIC")
print("------------------------")

print(
    "Ground-truth injected issues:",
    len(return_reason_ground_truth)
)

print(
    "Currently detected as invalid:",
    int(
        return_reason_diagnostic[
            "currently_invalid"
        ].sum()
    )
)

print(
    "Injected issues NOT currently detected:",
    int(
        (~return_reason_diagnostic[
            "currently_invalid"
        ]).sum()
    )
)

print("\nValid return reasons:")
print(sorted(valid_return_reasons))

print(
    "\nRows recorded as corrupted but currently appearing valid:"
)

display(
    return_reason_diagnostic.loc[
        ~return_reason_diagnostic[
            "currently_invalid"
        ]
    ].head(25)
)
```

```python
# ---------------------------------------------------------
# Step 146:
# Inspect Return-Reason Values for Undetected Issues
# ---------------------------------------------------------

undetected_return_reason_issues = (
    return_reason_diagnostic[
        ~return_reason_diagnostic["currently_invalid"]
    ]
    .copy()
)

print("Undetected injected return-reason issues:")
print(len(undetected_return_reason_issues))

print("\nColumns available in diagnostic table:")
print(undetected_return_reason_issues.columns.tolist())

display(
    undetected_return_reason_issues
)
```

```python
# ---------------------------------------------------------
# Step 147:
# Check Duplicate Return IDs in Undetected Reason Issues
# ---------------------------------------------------------

duplicate_return_ids = set(
    audit_returns.loc[
        audit_returns["return_id"].duplicated(keep=False),
        "return_id"
    ]
)

undetected_return_reason_issues[
    "is_duplicate_return_id"
] = (
    undetected_return_reason_issues[
        "return_id"
    ].isin(duplicate_return_ids)
)

print("RETURN-REASON DUPLICATE DIAGNOSTIC")
print("----------------------------------")

print(
    "Undetected return-reason issues:",
    len(undetected_return_reason_issues)
)

print(
    "Associated with duplicate return IDs:",
    int(
        undetected_return_reason_issues[
            "is_duplicate_return_id"
        ].sum()
    )
)

print(
    "Not associated with duplicate return IDs:",
    int(
        (~undetected_return_reason_issues[
            "is_duplicate_return_id"
        ]).sum()
    )
)

display(
    undetected_return_reason_issues[
        [
            "return_id",
            "return_reason",
            "is_duplicate_return_id"
        ]
    ].head(25)
)
```

```python
# ---------------------------------------------------------
# Step 148:
# Inspect Actual Return-Reason Injection Records
# ---------------------------------------------------------

print("Columns in return issue log:")
print(return_issue_log_df.columns.tolist())

print("\nReturn-reason issue records:")
display(
    return_reason_ground_truth.head(20)
)

print("\nUnique values in each useful issue-log column:")

for col in return_reason_ground_truth.columns:
    if col not in ["return_id"]:
        print(f"\n--- {col} ---")
        print(
            return_reason_ground_truth[col]
            .value_counts(dropna=False)
            .head(20)
        )
```

```python
# ---------------------------------------------------------
# Step 149:
# Compare Logged Return-Reason Value vs Final CSV Value
# ---------------------------------------------------------

possible_value_columns = [
    "injected_value",
    "new_value",
    "corrupted_value",
    "invalid_value"
]

injected_col = next(
    (
        col for col in possible_value_columns
        if col in return_reason_ground_truth.columns
    ),
    None
)

if injected_col is None:
    print("No injected-value column found automatically.")
    print(
        "Available columns:",
        return_reason_ground_truth.columns.tolist()
    )

else:
    comparison_df = (
        return_reason_ground_truth[
            ["return_id", injected_col]
        ]
        .merge(
            audit_returns[
                ["return_id", "return_reason"]
            ],
            on="return_id",
            how="left"
        )
    )

    comparison_df["logged_matches_final"] = (
        comparison_df[injected_col].astype(str)
        ==
        comparison_df["return_reason"].astype(str)
    )

    comparison_df["final_value_is_valid"] = (
        comparison_df["return_reason"]
        .isin(valid_return_reasons)
    )

    print("RETURN REASON VALUE COMPARISON")
    print("------------------------------")

    print(
        "Ground-truth issues:",
        len(comparison_df)
    )

    print(
        "Logged value matches final CSV:",
        int(
            comparison_df[
                "logged_matches_final"
            ].sum()
        )
    )

    print(
        "Logged value DOES NOT match final CSV:",
        int(
            (~comparison_df[
                "logged_matches_final"
            ]).sum()
        )
    )

    print(
        "Final values that appear valid:",
        int(
            comparison_df[
                "final_value_is_valid"
            ].sum()
        )
    )

    display(
        comparison_df[
            ~comparison_df["logged_matches_final"]
            |
            comparison_df["final_value_is_valid"]
        ].head(30)
    )
```

```python
# ---------------------------------------------------------
# Step 150:
# Validate Injected Return-Reason Values
# ---------------------------------------------------------

# Find the column that stores the injected/corrupted value
possible_value_columns = [
    "injected_value",
    "new_value",
    "corrupted_value",
    "invalid_value"
]

injected_col = next(
    (
        col for col in possible_value_columns
        if col in return_reason_ground_truth.columns
    ),
    None
)

if injected_col is None:

    print("Could not identify the injected-value column.")
    print("Available columns:")
    print(return_reason_ground_truth.columns.tolist())

else:

    reason_injection_check = (
        return_reason_ground_truth[
            ["return_id", injected_col]
        ]
        .copy()
    )

    reason_injection_check["injected_value_is_valid"] = (
        reason_injection_check[injected_col]
        .isin(valid_return_reasons)
    )

    print("RETURN-REASON INJECTION VALIDITY")
    print("--------------------------------")

    print(
        "Total logged return-reason issues:",
        len(reason_injection_check)
    )

    print(
        "Actually invalid injected values:",
        int(
            (~reason_injection_check[
                "injected_value_is_valid"
            ]).sum()
        )
    )

    print(
        "Injected values that are actually VALID:",
        int(
            reason_injection_check[
                "injected_value_is_valid"
            ].sum()
        )
    )

    print("\nInjected value distribution:")

    display(
        reason_injection_check[
            injected_col
        ]
        .value_counts(dropna=False)
        .rename_axis("injected_value")
        .reset_index(name="count")
    )

    print("\nLogged issues whose injected value is valid:")

    display(
        reason_injection_check[
            reason_injection_check[
                "injected_value_is_valid"
            ]
        ]
    )
```

```python
# ---------------------------------------------------------
# Step 151:
# Identify Root Cause of Return-Reason Discrepancy
# ---------------------------------------------------------

if injected_col is None:

    print("ROOT CAUSE CANNOT YET BE DETERMINED")
    print("No injected-value column was identified in the issue log.")

else:

    logged_valid_count = int(
        return_reason_ground_truth[injected_col]
        .isin(valid_return_reasons)
        .sum()
    )

    comparison = (
        return_reason_ground_truth[
            ["return_id", injected_col]
        ]
        .merge(
            audit_returns[
                ["return_id", "return_reason"]
            ],
            on="return_id",
            how="left"
        )
    )

    comparison["logged_matches_final"] = (
        comparison[injected_col].astype(str)
        ==
        comparison["return_reason"].astype(str)
    )

    changed_after_logging = int(
        (~comparison["logged_matches_final"]).sum()
    )

    detected_invalid = int(
        (~comparison["return_reason"].isin(
            valid_return_reasons
        )).sum()
    )

    print("RETURN-REASON ROOT-CAUSE SUMMARY")
    print("--------------------------------")

    print("Logged issues:", len(comparison))
    print("Detected invalid in final CSV:", detected_invalid)
    print("Logged injected values that were actually valid:", logged_valid_count)
    print("Values changed after issue logging:", changed_after_logging)

    print("\nDIAGNOSIS:")

    if logged_valid_count > 0:
        print(
            "The injection logic generated some values "
            "that belong to the valid return-reason domain."
        )

    elif changed_after_logging > 0:
        print(
            "The issue log recorded invalid values correctly, "
            "but some values changed before the final CSV was saved."
        )

    elif detected_invalid == len(comparison):
        print(
            "All logged return-reason issues are correctly "
            "present and detectable in the final dataset."
        )

    else:
        print(
            "The discrepancy requires additional investigation."
        )
```

```python
# ---------------------------------------------------------
# Step 152:
# Repair Return-Reason Corruption Injection
# ---------------------------------------------------------

# IDs that were intentionally selected for return_reason corruption
reason_issue_ids = (
    return_issue_log_df.loc[
        return_issue_log_df["column"]
        .astype(str)
        .str.lower()
        .eq("return_reason"),
        "return_id"
    ]
    .dropna()
    .unique()
)

# Values deliberately outside the valid business domain
invalid_reason_values = [
    "Unknown Reason",
    "Other Invalid",
    "System Error",
    "Invalid Code"
]

# Work from current dirty returns dataset
returns_repaired_df = audit_returns.copy()

# Do not modify duplicate copies separately
reason_mask = (
    returns_repaired_df["return_id"]
    .isin(reason_issue_ids)
)

# Deterministic replacement
rng = np.random.default_rng(RANDOM_SEED)

returns_repaired_df.loc[
    reason_mask,
    "return_reason"
] = rng.choice(
    invalid_reason_values,
    size=reason_mask.sum()
)

print("Intended return_reason issue IDs:", len(reason_issue_ids))
print("Rows updated:", int(reason_mask.sum()))
```

```python
# ---------------------------------------------------------
# Step 153:
# Verify Repaired Return-Reason Issues
# ---------------------------------------------------------

# Derive valid reasons from the clean ground-truth dataset
valid_return_reasons = set(
    returns_ground_truth_df[
        "return_reason"
    ].dropna().unique()
)

# Detect invalid reasons in repaired dataset
invalid_reasons_after_repair = (
    returns_repaired_df[
        returns_repaired_df["return_reason"].notna()
        &
        ~returns_repaired_df[
            "return_reason"
        ].isin(valid_return_reasons)
    ]
    .copy()
)

expected_reason_issues = len(reason_issue_ids)
detected_reason_issues = len(invalid_reasons_after_repair)

print("RETURN-REASON REPAIR VERIFICATION")
print("---------------------------------")

print(
    "Expected invalid return reasons:",
    expected_reason_issues
)

print(
    "Detected invalid return reasons:",
    detected_reason_issues
)

print(
    "\nPASS"
    if detected_reason_issues == expected_reason_issues
    else "\nREVIEW"
)

print("\nInvalid reason distribution:")

display(
    invalid_reasons_after_repair[
        "return_reason"
    ]
    .value_counts()
    .rename_axis("invalid_reason")
    .reset_index(name="count")
)
```
